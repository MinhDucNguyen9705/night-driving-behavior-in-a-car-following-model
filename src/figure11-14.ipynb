```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
from helper_functions_v2 import plot_raster
```

```python
from helper_functions_v2 import compute_fundamental_diagram, optimal_velocity, optimal_velocity_normal

N_start=50
N_end=450
N_step=10
L=500
N_values = np.arange(N_start, N_end + 1, N_step)
densities = N_values / L

normal_flows = densities * np.array([optimal_velocity_normal(L / N) for N in N_values])
night_flows = densities * np.array([optimal_velocity(L / N) for N in N_values])
```

## fig 11

```python
fig, ax = plot_raster(N=220, n_dec=80, kappa=1, lambda_=0.5)
fig.savefig('../figures/figure_11_reproduction.png', dpi=300)
```

## fig 12

```python
densities, flows = compute_fundamental_diagram(
    kappa=1, 
    lambda_=0.2, 
    n_dec=80,
    L=L,
    N_start=N_start,
    N_end=N_end,
    N_step=N_step
)

plt.figure(figsize=(8, 5))
plt.plot(densities, normal_flows, '-', label="Normal driving")
plt.plot(densities, night_flows, '--', label="Night driving without perturbations")
plt.plot(densities, flows, ':', linewidth=2.0, label="Night driving with perturbations")

plt.xlabel("Density (ρ = N / L)")
plt.ylabel("Flow (q = ρ · v̄)")
plt.title("Fig 12")
plt.grid(True)
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig('../figures/figure_12_reproduction.png', dpi=300)
plt.show()
```
